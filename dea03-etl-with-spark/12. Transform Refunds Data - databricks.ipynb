# Transform Refunds Data
1.Extract Specific Portions of the string from refunds_reason using split function.

2.Extract Specific Portions of the string from refunds_reason using regexp_extract function.

3.Extract Date and time from the refund_timestamp.

4.Write Transformed data to Silver Schema in hive_metastore.


# 1.Extract Specific Portions of the string from refunds_reason using split function.

```sql
select  refund_id,
 payment_id,
 CAST(date_format(refund_timestamp, 'yyyy-MM-dd') as date) as refund_date,
 date_format(refund_timestamp, 'HH:mm:ss') as refund_time,
 refund_amount,
 split(refund_reason,':')[0] as refund_reason,
 split(refund_reason,':')[1] as refund_source 
from hive_metastore.bronze.refunds;
```

```sql
select  refund_id,
 payment_id,
 refund_timestamp,
 refund_amount,
 regexp_extract(refund_reason, '^([^:]+):',1 ) as refund_reason,
 regexp_extract(refund_reason, '^([^:]+):(.*)$',2 ) as refund_source 
from hive_metastore.bronze.refunds;
```

# 3.Extract Date and time from the refund_timestamp.


```sql
CREATE OR REPLACE TEMPORARY VIEW refunds 
AS
select  refund_id,
 payment_id,
 CAST(date_format(refund_timestamp, 'yyyy-MM-dd') as date) as refund_date,
 date_format(refund_timestamp, 'HH:mm:ss') as refund_time,
 refund_amount,
 split(refund_reason,':')[0] as refund_reason,
 split(refund_reason,':')[1] as refund_source 
from hive_metastore.bronze.refunds;
```

```sql
select * from refunds;
```

# 4.Write Transformed data to Silver Schema in hive_metastore.

```sql
CREATE SCHEMA IF NOT EXISTS hive_metastore.silver;
```

```sql
CREATE TABLE IF NOT EXISTS hive_metastore.silver.refunds 
AS
SELECT * FROM refunds;
```
