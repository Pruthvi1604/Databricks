# 2. Create External Location

```sql
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricksdemo001_gizmobox
URL 'abfss://gizmobox@databricksdemo001.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL dea_course_ext_sc)
COMMENT 'Apache Spark ETL project'
```

```sql
%fs ls 'abfss://gizmobox@databricksdemo001.dfs.core.windows.net/'
```

```sql
%fs ls 'abfss://gizmobox@databricksdemo001.dfs.core.windows.net/landing'
```

# 3. Create the catalog - gizmobox

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS gizmobox
MANAGED LOCATION 'abfss://gizmobox@databricksdemo001.dfs.core.windows.net'
COMMENT 'creating gizmobox catalog';
```

```sql
SHOW CATALOGS;
```

# 5. Create Volumes

```sql
USE CATALOG gizmobox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operation_data
LOCATION 'abfss://gizmobox@databricksdemo001.dfs.core.windows.net/landing/operational_data/'
COMMENT 'creating volumes for operational data';
```

```sql
show volumes;
```
