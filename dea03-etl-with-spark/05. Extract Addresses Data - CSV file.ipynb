# Extract Addresses Data from the Address Files
#### 1. Demonstrate Limitations of CSV file format in SELECT statement
#### 2. Use read_files Function to overcome the limitations
#### 3. Create Addresses view in Bronze Layer

##1. Demonstrate Limitations of CSV file format in SELECT statement

```sql
select *
from csv.`/Volumes/gizmobox/landing/operation_data/addresses/*`
```

## 2. Use read_files Function to overcome the limitations

```sql
select *
from read_files('/Volumes/gizmobox/landing/operation_data/addresses/',
                format => 'csv',
                header => true,
                delimiter => '\t')
```

## 3. Create Addresses view in Bronze Layer


```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_addresses
AS
select *
from read_files('/Volumes/gizmobox/landing/operation_data/addresses/',
                format => 'csv',
                header => true,
                delimiter => '\t')
```

```sql
select * from gizmobox.bronze.v_addresses
```
