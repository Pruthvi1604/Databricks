```sql
SELECT *
FROM gizmobox.bronze.v_orders;
```

#1. Extract Top Level Column Values
Top Level Object <column_name>:<extract_path>

```sql
SELECT value: order_id as order_id,
 value
FROM gizmobox.bronze.v_orders;
```

#2. Extract Array Elements
To Extract array elements - use the indices in brackets <columne_name>:<extract_path>[index]

```sql
SELECT value: items[0] as items_1,
 value: items[1] as items_2,
 value
FROM gizmobox.bronze.v_orders;
```
