# Querying the files from customers folder in "landing" schema and "operational_data" volume

### example: select * from <file_format>.`<file_path>`

# 1. Querying Single json file

```sql
select * 
from json.`/Volumes/gizmobox/landing/operation_data/customers/customers_2024_10.json`
```

# 2. Querying using wild card characters

```sql
select * 
from json.`/Volumes/gizmobox/landing/operation_data/customers/customers_2024_*`
```

# 3. Querying entire folder

```sql
select *
from json.`/Volumes/gizmobox/landing/operation_data/customers/*`
```

# 4. Select File metadata

```sql
select input_file_name() as file_path, --Depricated from databricks Runtime 13.3 LTS onwards
*
from json.`/Volumes/gizmobox/landing/operation_data/customers/`
```

```sql
select input_file_name() as file_path, 
_metadata.file_path as file_path_new, 
*
from json.`/Volumes/gizmobox/landing/operation_data/customers/`
```

# 5. Register Files in Unity Catalog using Views

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers
AS 
select _metadata.file_path as file_path_new, 
*
from json.`/Volumes/gizmobox/landing/operation_data/customers/`
```

```sql
  select * from gizmobox.bronze.v_customers
```

# 6. Create Temporary view 

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_customers
AS 
select _metadata.file_path as file_path_new, 
*
from json.`/Volumes/gizmobox/landing/operation_data/customers/`
```

# 7. Create Global Temporary view

```sql
CREATE OR REPLACE GLOBAL TEMPORARY VIEW gtv_customers
AS 
select _metadata.file_path as file_path_new, 
*
from json.`/Volumes/gizmobox/landing/operation_data/customers/`
```
